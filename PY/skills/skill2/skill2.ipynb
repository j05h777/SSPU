## Question 1
**Problem Statement:** Create a warehouse management system to add, delete, search, and display products through a menu-driven interface.

**Algorithm:**
1. Start
2. Create an empty inventory dictionary.
3. Display the menu and read the user's choice.
4. Perform the selected add, delete, search, or display operation.
5. Show an error for an invalid choice.
6. Repeat until Exit is selected.
7. End

**Flowchart:**
```mermaid
flowchart TD
    A([Start]) --> B[Create empty inventory]
    B --> C[Display menu and read choice]
    C --> D{Which option?}
    D -->|Add| E[Add product]
    D -->|Delete| F[Delete product]
    D -->|Search| G[Search product]
    D -->|Display| H[Display inventory]
    D -->|Invalid| I[Show error]
    D -->|Exit| J([End])
    E --> C
    F --> C
    G --> C
    H --> C
    I --> C
```

```mermaid
        string name

    }
```
    INVENTORY {

`


````python
#Skill 2:
#Q.1 Warehouse Management
# Warehouse Management System

inventory = {}

def add_product():
    product_id = input("Enter Product ID: ")

    if product_id in inventory:
        print("Product already exists.")
        return

    name = input("Enter Product Name: ")
    category = input("Enter Category: ")

    try:
        quantity = int(input("Enter Quantity: "))
        price = float(input("Enter Unit Price: "))

        if quantity < 0 or price < 0:
            print("Values cannot be negative.")
            return

    except ValueError:
        print("Please enter valid numeric values.")
        return

    inventory[product_id] = {
        "name": name,
        "category": category,
        "quantity": quantity,
        "price": price,
    }

    print("Product added successfully.")
'''
def update_product():
    product_id = input("Enter Product ID to update: ")

    if product_id not in inventory:
        print("Product not found.")
        return

    print("\n1. Update Name")
    print("2. Update Category")
    print("3. Update Price")

    choice = input("Enter your choice: ")

    if choice == "1":
        inventory[product_id]["name"] = input("Enter new name: ")

    elif choice == "2":
        inventory[product_id]["category"] = input("Enter new category: ")

    elif choice == "3":
        try:
            price = float(input("Enter new price: "))

            if price < 0:
                print("Price cannot be negative.")
                return

            inventory[product_id]["price"] = price

        except ValueError:
            print("Invalid price.")
            return
        
    else:
        print("Invalid choice.")
        return

    print("Product updated successfully.")
'''
def delete_product():
    product_id = input("Enter Product ID to delete: ")

    if product_id not in inventory:
        print("Product not found.")
        return

    del inventory[product_id]

    print("Product deleted successfully.")

def search_product():
    product_id = input("Enter Product ID to search: ")

    if product_id not in inventory:
        print("Product not found.")
        return

    product = inventory[product_id]

    print("\n--- Product Details ---")
    print("Product ID     :", product_id)
    print("Product Name   :", product["name"])
    print("Category       :", product["category"])
    print("Quantity       :", product["quantity"])
    print("Unit Price     : ₹", product["price"])

def display_products():
    if not inventory:
        print("Inventory is empty.")
        return

    print("\n\t\tINVENTORY")
    print(
        f"{'ID':<10}"
        f"{'Name':<20}"
        f"{'Category':<15}"
        f"{'Quantity':<10}"
        f"{'Price':<12}"
    )

    for product_id, product in inventory.items():
        print(
            f"{product_id:<10}"
            f"{product['name']:<20}"
            f"{product['category']:<15}"
            f"{product['quantity']:<10}"
            f"₹{product['price']:<11.2f}"
        )

def main():

    while True:

        print("\n")
        print("\t\tWAREHOUSE MANAGEMENT SYSTEM")
        print("1. Add Product")
        #print("2. Update Product")
        print("2. Delete Product")
        print("3. Search Product")
        print("4. Display Inventory")
        print("5. Exit")

        choice = input("Enter your choice: ")
        if choice == "1":
            add_product()
        # elif choice == "2":
        #     update_product()
        elif choice == "2":
            delete_product()
        elif choice == "3":
            search_product()
        elif choice == "4":
            display_products()
        elif choice == "5":
            print("Exiting...")
            break
        else:
            print("Invalid choice. Please try again.")

# Program execution
main()
````

## Question 2
**Problem Statement:** Analyze student marks using list operations to find statistics, sort marks, count high scores, add a mark, and remove the lowest mark.

**Algorithm:**
1. Start
2. Create a list of student marks.
3. Display the marks and calculate maximum, minimum, and average.
4. Display ascending and descending orders.
5. Count students scoring above 75.
6. Append 100 and remove the lowest mark.
7. Display the updated list.
8. End

**Flowchart:**
```mermaid
flowchart TD
    A([Start]) --> B[Create marks list]
    B --> C[Calculate maximum, minimum, and average]
    C --> D[Sort marks]
    D --> E[Count marks above 75]
    E --> F[Append 100]
    F --> G[Remove lowest mark]
    G --> H[/Display results/]
    H --> I([End])
```


**ER Diagram:**
```mermaid
erDiagram
    STUDENT ||--o{ MARKS : records

    STUDENT {
        int roll_no PK
        string name
        int total_marks
    }

    MARKS {
        int student_id FK
        int mark_value
    }
```

````python
#Skill 2: Student marks analysis using lists
#Q.2 WAP to perform various list operations on the students marks

marks=[56,78,90,45,67,89,76,54,32,98]

print("Marks of all students:",marks)
print("Maximum marks:",max(marks))
print("Minimum marks:",min(marks))
print("Average marks:",sum(marks)/len(marks))

print("Sorted marks (ascending):",sorted(marks))

# marks1=marks.copy()
# marks1.reverse()
# print("Sorted marks (descending):",marks1)
print("Sorted marks (descending):",sorted(marks,reverse=True))

print("No of student scoring above 75:",len([x for x in marks if x>75]))

marks.append(100)
print("Updated marks:",marks)

marks.remove(min(marks))
print("Updated marks after removing lowest marks:",marks)
````

## Question 4 - ALTERNATIVE

**Algorithm:**
1. Start
2. Set the student records filename.
3. Display the menu and read the user's choice.
4. Add, display, search, or delete records according to the choice.
5. Handle missing files and invalid choices.
6. Repeat until Exit is selected.
7. End

**Flowchart:**
```mermaid
flowchart TD
    A([Start]) --> B[Set records filename]
    B --> C[Display menu and read choice]
    C --> D{Which option?}
    D -->|Add| E[Write student record]
    D -->|Display| F[Read all records]
    D -->|Search| G[Find record by roll number]
    D -->|Delete| H[Remove record by roll number]
    D -->|Invalid| I[Show error]
    D -->|Exit| J([End])
    E --> C
    F --> C
    G --> C
    H --> C

```


**ER Diagram:**
```mermaid

    STUDENT {
```

````python
# Skill 2:
# Q.3: (ALTERNATIVE) File Handling – Student Record
# Student Record Management System using File Handling

filename = "students.txt"

while True:
    print("\n\t\tSTUDENT RECORD SYSTEM")
    print("1. Add Student")
    print("2. Display Students")
    print("3. Search Student")
    #print("4. Update Student")
    print("4. Delete Student")
    print("5. Exit")
    choice = input("Enter your choice: ")

    # Add Student
    if choice == "1":
        roll_no = input("Enter Roll Number: ")
        name = input("Enter Name: ")
        course = input("Enter Course: ")
        marks = input("Enter Marks: ")
        with open(filename, "a") as file:
            file.write(f"{roll_no},{name},{course},{marks}\n")
        print("Student record added successfully.")

    # Display Students
    elif choice == "2":
        try:
            with open(filename, "r") as file:
                print("\n\t\tStudent Records")
                found = False
                for line in file:
                    roll_no, name, course, marks = line.strip().split(",")
                    print(
                        "Roll No:", roll_no,
                        "| Name:", name,
                        "| Course:", course,
                        "| Marks:", marks
                    )
                    found = True
                if not found:
                    print("No student records found.")
        except FileNotFoundError:
            print("No student records found.")

    # Search Student
    elif choice == "3":
        search = input("Enter Roll Number to search: ")
        try:
            with open(filename, "r") as file:
                found = False
                for line in file:
                    roll_no, name, course, marks = line.strip().split(",")
                    if roll_no == search:
                        print("\nStudent Found")
                        print("Roll No :", roll_no)
                        print("Name    :", name)
                        print("Course  :", course)
                        print("Marks   :", marks)
                        found = True
                        break
                if not found:
                    print("Student not found.")
        except FileNotFoundError:
            print("No student records found.")
    
    # Delete Student
    elif choice == "4":
        search = input("Enter Roll Number to delete: ")
        try:
            found = False
            with open(filename, "wr") as file:
                lines = file.readlines()
                for line in lines:
                    roll_no = line.strip().split(",")[0]
                    if roll_no == search:
                        found = True
                    else:
                        file.write(line)
            if found:
                print("Student record deleted successfully.")
            else:
                print("Student not found.")
        except FileNotFoundError:
            print("No student records found.")
    elif choice == "5":
        print("Exiting...")
        break
    else:
        print("Invalid choice. Please try again.")

'''
    # Update Student
    elif choice == "4":

        search = input("Enter Roll Number to update: ")

        try:
            with open(filename, "r") as file:
                lines = file.readlines()

            found = False

            with open(filename, "w") as file:

                for line in lines:

                    roll_no, name, course, marks = line.strip().split(",")

                    if roll_no == search:

                        print("Enter new details:")
                        name = input("Enter Name: ")
                        course = input("Enter Course: ")
                        marks = input("Enter Marks: ")
                        file.write(
                            f"{roll_no},{name},{course},{marks}\n"
                        )
                        found = True
                    else:
                        file.write(line)
            if found:
                print("Student record updated successfully.")
            else:
                print("Student not found.")
        except FileNotFoundError:
            print("No student records found.")
    '''
````
