## Q4.1: Cybersecurity Login Audit

Case study: 
- A security team wants to identify accounts with at least three failed login attempts. 
- Usernames may have inconsistent capitalization. 
- Passwords must not be stored in the audit report.

Task: Count failed attempts for each normalized username and flag suspicious accounts.

```python
login_records = [
    ("John", "pass123", "failed"),
    ("ALICE", "abc123", "success"),
    ("john", "xyz456", "failed"),
    ("Bob", "hello123", "failed"),
    ("JOHN", "test123", "failed"),
    ("bob", "pass456", "failed"),
    ("Alice", "pwd123", "failed")
]

# Dictionary to count failed attempts
failed_attempts = {}

for username, password, status in login_records:
    username = username.strip().lower()
    if status == "failed":
        failed_attempts[username] = failed_attempts.get(username, 0) + 1

# Identify suspicious accounts
suspicious_accounts = {}

for user, count in failed_attempts.items():
    if count >= 3:
        suspicious_accounts[user]=count

print("Failed Login Attempts:", failed_attempts)
print("Suspicious Accounts:", suspicious_accounts)
```

## Q4.2: Clinic Appointment Deduplication


Case study: 
- A clinic's appointment export contains duplicate appointment references. 
- The reporting team must remove duplicates without changing the original order of first occurrence.

```python
appointments = [
    "APT101", "APT102", "APT103",
    "APT101", "APT104", "APT102",
    "APT105", "APT103"
]

# Remove duplicates while preserving order
unique_appointments = list(dict.fromkeys(appointments))

print("Original Appointments:", appointments)
print("Unique Appointments:", unique_appointments)
print("Total Unique Appointments:", len(unique_appointments))
```
