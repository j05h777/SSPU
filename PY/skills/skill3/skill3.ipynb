```python
# S3Q1) 
# a) Design a NumPy representation for a 30-day × 24-hour ×10-sensor dataset.
# b) Suppose: 1000 observations × 20 sensors. Determine: appropriate NumPy representation, shape, extraction of one sensor, extraction of one observation, and statistical operations.
# c) Given a (10000, 50) matrix, explain how to extract one feature and one observation efficiently.

import numpy as np
data=np.random.rand(1000,20)
print("data:\n",data[0:5,0:5])  
print("\ndata shape:", data.shape)
print("data type:", data.dtype)
print("data size:", data.size)

# Extraction of one sensor (1000 observations)
sensor_1 = data[:, 0]
print("\nSensor 1 data shape:", sensor_1.shape)

# Extraction of one observation (20 sensors)
observation_1 = data[0, :]
print("Observation 1 data shape:", observation_1.shape)

# Statistical operations
print("\nMean :", round(np.mean(data), 2))
print("Standard deviation:", round(np.std(data), 2))
print("Mean of sensor 1:", round(np.mean(sensor_1), 2))
print("Standard deviation of sensor 1:", round(np.std(sensor_1), 2))
```

```python
# S3Q2) Create Using Series and Data Frame . 
# Assume that you have 1000 students records of MSc Data Science
# a) Student Performance Analytics. Identify:highest performer, average marks, students below threshold, attendance-risk students.
# b) Analyze a student-performance DataFrame.
# c) Build a DataFrame-based sales analysis.

# S3Q2a)
import pandas as pd
students = pd.DataFrame({
    "student_id": range(1, 1001),
    "age": np.random.randint(20, 35, 1000),
    "gender": np.random.choice(["M","F"], 1000),
    "marks": np.random.randint(0, 101, 1000),
    "attendance": np.random.randint(0, 101, 1000)
})

highest_performer = students.loc[students['marks'].idxmax()]
average_marks = students['marks'].mean()
students_below_threshold = students[students['marks']<50]
attendance_risk_students = students[students['attendance']<75]

print("Highest Performer:")
print(highest_performer)
print("\nAverage Marks:", round(average_marks, 2))
print("Number of Students Below Threshold:", len(students_below_threshold))
print("Number of Attendance Risk Students:", len(attendance_risk_students))
```

```python
# S3Q2b)
print("\nStudent DataFrame Analysis:")

print("\nShape of DataFrame:", students.shape)

print("\nFirst 5 Students:")
print(students.head())

print("\nSummary Statistics:")
print(students.describe())

# Average marks by gender
average_marks_gender = students.groupby("gender")["marks"].mean()

print("\nAverage Marks by Gender:")
print(average_marks_gender.round(2))

# Average attendance by gender
average_attendance_gender = students.groupby("gender")["attendance"].mean()

print("\nAverage Attendance by Gender:")
print(average_attendance_gender.round(2))

# Count of students by gender
gender_count = students["gender"].value_counts()

print("\nNumber of Students by Gender:")
print(gender_count)
```

```python
# S3Q2c)

sales = pd.DataFrame({
    "product": ["Laptop", "Mouse", "Keyboard", "Monitor", "Laptop",
                "Mouse", "Monitor", "Keyboard", "Laptop", "Monitor"],
    "category": ["Electronics", "Accessories", "Accessories", "Electronics",
                 "Electronics", "Accessories", "Electronics", "Accessories",
                 "Electronics", "Electronics"],
    "quantity": [5, 20, 15, 8, 3, 25, 6, 18, 4, 10],
    "price": [60000, 800, 1500, 12000, 60000, 800, 12000, 1500, 60000, 12000]
})

# Calculate revenue
sales["revenue"] = sales["quantity"] * sales["price"]

print("\nSales Data:")
print(sales)

# Total revenue
total_revenue = sales["revenue"].sum()
print("\nTotal Revenue:", total_revenue)

# Total quantity sold by product
product_quantity = sales.groupby("product")["quantity"].sum()

print("\nQuantity Sold by Product:")
print(product_quantity)

# Revenue by product
product_revenue = sales.groupby("product")["revenue"].sum()

print("\nRevenue by Product:")
print(product_revenue)

# Revenue by category
category_revenue = sales.groupby("category")["revenue"].sum()

print("\nRevenue by Category:")
print(category_revenue)

# Best-selling product
best_selling_product = product_quantity.idxmax()

print("\nBest-Selling Product:", best_selling_product)
```
