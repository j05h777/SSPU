```python
# Practical 2: Tuple operations for product information
# to use tuples to store immutable product info 
# create tuples containing product id,name,price

# tuple packing
product = (101, "Laptop", 50000)
print("Packed tuple:", product)

# tuple indexing
print("Product ID:", product[0])
print("Product Name:", product[1])
print("Product Price:", product[2])

# tuple unpacking
pid, pname, pprice = product
print("Unpacked tuple:", pid, pname, pprice)

#tuple concatenation
product2 = (102, "Smartphone", 30000)
product3 = (103, "Laptop", 45000)
combined_products = product + product2 + product3
print("Combined Products:", combined_products)

# count occurence of a value
print("Count of laptop:",combined_products.count('Laptop'))

# slicing
print("Sliced Tuple:", combined_products[0:3])

# indexing
print("Index of pid 103:",combined_products.index(103))


# immutability demo
try:
    product[2] = 55000
except TypeError as e:
    print("\nTuple is immutable:", e)

```

```python
# sets
set1={10,20,30,40,50}
set2={30,40,50,60,70}

# union
print("Union:",set1.union(set2))
print("Union:",set1|set2)

# intersection
print("Intersection:",set1.intersection(set2))
print("Intersection:",set1&set2)

# difference
print("Difference:",set1.difference(set2))
print("Difference:",set1-set2)

# symmetric difference
print("Symmetric difference:",set1.symmetric_difference(set2))
print("Symmetric difference:",set1^set2)
# 
```

```python
# Practical 3: Set operations on customer ids
# objective: perform operations to identify common and unique customers
# question: given 2 sets of customer ids from 2 stores, perform: 
# union, intersection, difference, symmetric difference, add and remove an element

s1 = {101, 102, 103, 104, 105}
s2 = {104, 105, 106, 107, 108}

print("Union:", s1.union(s2))

print("Intersection:", s1.intersection(s2))

print("Difference:", s1.difference(s2))

print("Symmetric Difference:", s1.symmetric_difference(s2))

s1.add(109)
s1.remove(101)
print("Updated Store 1:", s1)
```

```python
# Practical 4: Customer database using dictionaries
# objective: store customer information using dictionaries
# question: create a dictionary containing customer details. 
# perform: add new customer, update age, delete a customer, display all keys and values, iterate through the dictionary

cust_db={
    "Name": ["John", "Jason", "Rachel"],
    "Age":[20,21,25]
}
print(cust_db)

# add
cust_db["Name"].append("Monica")
cust_db["Age"].append(29)
print(cust_db)

# update
cust_db["Age"][0]=30
print(cust_db)

# delete
#cust_db["ID"].remove(104)
#cust_db["Name"].remove("Monica")
#cust_db["Age"].remove(29)
del cust_db["Name"][2]
del cust_db["Age"][2]
print(cust_db,"\n")

# display all keys and values 
print("Keys:", cust_db.keys())
print("Values:", cust_db.values(),"\n")

# iteration
for key, values in cust_db.items():
    print(key, values)

#for name, age in zip(cust_db["Name"], cust_db["Age"]):
#   print(id, name, age)
```

```python
# practical 5: String processing
# objective: Perform different string manipulation operations
# question: input a string and perform: 
# convert to uppercase, lowercase, count vowels, reverse the string, replace one word, split into words

strinput=input("Enter a string:")
print(strinput.upper())
print(strinput.lower())
print("Vowel count:", sum(1 for v in strinput if v.lower() in 'aeiou'))
print("Reversed string:", strinput[::-1])
print("Replaced string:", strinput.replace('hello', 'hi'))
print("Split:", strinput.split())
```
