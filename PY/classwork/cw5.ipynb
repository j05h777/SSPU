```python
class bankaccount:
    def __init__(self,balance):
        self.balance=balance

    def deposit(self,amount):
        self.balance+=amount

    def withdraw(self,amount):
        if amount>self.balance:
            print("Insufficient balance")
        else:
            self.balance-=amount

    def display(self):
        print(self.balance)

b1=bankaccount(10000)
b1.deposit(2000)
b1.withdraw(4000)
b1.display()
```

```python
#p3q1
#create a product class with: product name,price,quantity,method to calculate total value
class product():
    def __init__(self,name,price,quantity):
        self.name=name
        self.price=price
        self.quantity=quantity

    def total_value(self):
        return self.price*self.quantity

p=product("laptop",10000,2)
p.total_value()
```

```python
# create a class people with some basic attributes and behaviour(speak,walk,breathe,aging) 
# and inherited class student and add behaviours(studentid,prn,course,last degree,year of enrollment). add some behaviour(attend exam,go for class,participate in events,)
# now create an employee class which inherits the student class (eid,department,designation,salary). add some behaviour(submit report,calc salary...)
# now create 2 object of people,3 of student,5 of employee
# access attribute of people and employee through student object

class people():
    def __init__(self,name,address,email,gender,age):
        self.name = name
        self.address = address
        self.email = email
        self.gender = gender
        self.age = age

    def speak(self):
        print(f"{self.name} is speaking.")

    def walk(self):
        print(f"{self.name} is walking.")

    def breathe(self):
        print(f"{self.name} is breathing.")

    def age(self):
        print(f"{self.name} is aging.")

class student(people):
    def __init__(self, name, address, email, gender, age, studentid, prn, course, last_degree, year_of_enrollment):
        super().__init__(name, address, email, gender, age)
        self.studentid = studentid
        self.prn = prn
        self.course = course
        self.last_degree = last_degree
        self.year_of_enrollment = year_of_enrollment

    def attend_exam(self):
        print(f"{self.name} is attending an exam.")

    def go_for_class(self):
        print(f"{self.name} is going for class.")

    def participate_in_events(self):
        print(f"{self.name} is participating in events.")

class employee(student):
    def __init__(self, name, address, email, gender, age, studentid, prn, course, last_degree, year_of_enrollment, eid, department, designation, salary):
        super().__init__(name, address, email, gender, age, studentid, prn, course, last_degree, year_of_enrollment)
        self.eid = eid
        self.department = department
        self.designation = designation
        self.salary = salary

    def submit_report(self):
        print(f"{self.name} is submitting a report.")

    def calc_salary(self):
        print(f"{self.name} is calculating salary.")
        print(self.salary * 12)

people1 = people("Alice", "Goa", "alice@example.com", "Female", 25)
people2 = people("Bob", "Mumbai", "bob@example.com", "Male", 30)

student1 = student("Charlie", "Delhi", "charlie@example.com", "Male", 20, 1, 101, "Computer Science", "Bachelor", 2020)
student2 = student("David", "Kolkata", "david@example.com", "Male", 22, 2, 102, "Mechanical Engineering", "Bachelor", 2020)
student3 = student("Eve", "Chennai", "eve@example.com", "Female", 21, 3, 103, "Electrical Engineering", "Bachelor", 2020)

employee1 = employee("Frank", "Bangalore", "frank@example.com", "Male", 35, 4, 104, "Civil Engineering", "Master", 2020, 1001, "IT", "Software Engineer", 50000)
employee2 = employee("Grace", "Hyderabad", "grace@example.com", "Female", 32, 5, 105, "Chemical Engineering", "Master", 2020, 1002, "HR", "HR Manager", 60000)
employee3 = employee("Henry", "Pune", "henry@example.com", "Male", 38, 6, 106, "Biomedical Engineering", "Master", 2020, 1003, "Finance", "Financial Analyst", 55000)
employee4 = employee("Ivy", "Jaipur", "ivy@example.com", "Female", 34, 7, 107, "Environmental Engineering", "Master", 2020, 1004, "Marketing", "Marketing Manager", 65000)
employee5 = employee("Jack", "Lucknow", "jack@example.com", "Male", 40, 8, 108, "Nuclear Engineering", "Master", 2020, 1005, "Operations", "Operations Manager", 70000)

student1.speak()
student1.go_for_class()
#student1.submit_report()

employee1.speak()
employee1.submit_report()
```

```python
# s3q1
# case study: IoT sensor 
# suppose: 1000 observations x 20 sensors
# determine: appropriate numpy representation,shape,extraction of one sensor, one observation, statistical operations on the data

data=np.random.rand(1000,20)
print("data:\n",data[0:5,0:5])  
print("\ndata shape:", data.shape)
print("data type:", data.dtype)
print("data size:", data.size)

# Extraction of one sensor (1000 observations)
sensor_1 = data[:, 0]
print("\nSensor 1 data shape:", sensor_1.shape)

# Extraction of one observation (20 sensors)
observation_1 = data[0, :]
print("Observation 1 data shape:", observation_1.shape)

# Statistical operations
print("\nMean :", round(np.mean(data), 2))
print("Standard deviation:", round(np.std(data), 2))
print("Mean of sensor 1:", round(np.mean(sensor_1), 2))
print("Standard deviation of sensor 1:", round(np.std(sensor_1), 2))
```

```python
# case study: retail pricing
# 1000 products x 5 stores
# Each store has a different discount
# final_price=prices*(1-discount)
# Vectorized retail pricing: 1000 products x 5 stores
import numpy as np

prices = np.random.randint(100, 10000, size=(1000, 5))
discounts = np.array([0.05, 0.10, 0.15, 0.20, 0.25])

# Broadcasting applies each store's discount to its entire column.
final_prices = prices * (1 - discounts)

print("Prices shape:", prices.shape)
print("Discounts shape:", discounts.shape)
print("Final prices shape:", final_prices.shape)
print("Final prices for the first 5 products:\n", final_prices[:5])
```
