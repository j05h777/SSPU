```python
l=list([100,101,100,102,105,99,86])
#indexing
print(l[2])

#negative indexing
print(l[-2])

#slicing
print(l[0:4])

#traversing
for i in l:
    print(i,end=" ")

#concatenation
l1=[1,2,3]
l2=[4,5,6]
l3=l1+l2
print("\n",l1,l2,l3)

#repetition
l4=l1*3
print(l4)
```

```python
l5=[1,2,3,4,5]
print(l5)

#append
l5.append(6)
print(l5)

#insert
l5.insert(2,6)
print(l5)

#extend
l5.extend([7,8,9])
print(l5)

#remove
l5.remove(6)
print(l5)

#pop
l5.pop()
print(l5)

#reverse
l5.reverse()
print(l5)

#sort
l5.sort()
print(l5)

#copy
l6=l5.copy()
#sort
l5.sort()
print(l6)

```

```python
#list comprehension
sq=[x**2 for x in range(1,6)]
print(sq)
```

```python
#Practical 1: Student marks analysis using lists
#Q.1 WAP to perform various list operations on the students marks

marks=[56,78,90,45,67,89,76,54,32,98]

print("Marks of all students:",marks)
print("Maximum marks:",max(marks))
print("Minimum marks:",min(marks))
print("Average marks:",sum(marks)/len(marks))

print("Sorted marks (ascending):",sorted(marks))

marks1=marks.copy()
marks1.reverse()
print("Sorted marks (descending):",marks1)
#print("Sorted marks (descending):",sorted(marks,reverse=True))

print("No of student scoring above 75:",len([x for x in marks if x>75]))

marks.append(100)
print("Updated marks:",marks)

marks.remove(min(marks))
print("Updated marks after removing lowest marks:",marks)
```
