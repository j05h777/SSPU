## Question 1
**Problem Statement:** Generate all possible lists `[i, j, k]` such that `0 <= i <= x`, `0 <= j <= y`, `0 <= k <= z`, and `i + j + k != n`.

**Algorithm:**
1. Start
2. Read `x`, `y`, `z`, and `n`.
3. Use nested iteration to form every possible combination of `i`, `j`, and `k`.
4. Keep only those combinations whose sum is not equal to `n`.
5. Store the valid combinations in a list.
6. Display the final list.
7. End

**Flowchart:**
```mermaid
flowchart TD
    A([Start]) --> B[/Input x, y, z, n/]
    B --> C[Generate all i, j, k combinations]
    C --> D{Is i + j + k != n?}
    D -->|Yes| E[Add list item]
    D -->|No| F[Skip item]
    E --> G[/Display result/]
    F --> G
    G --> H([End])
```
**ER Diagram:**
```mermaid
erDiagram
    INPUT_PARAMETERS ||--o{ COMBINATION : generates
    INPUT_PARAMETERS {
        int parameter_id PK
        int x_limit
        int y_limit
        int z_limit
        int excluded_sum
    }
    COMBINATION {
        int combination_id PK
        int parameter_id FK
        int i_value
        int j_value
        int k_value
        int total
    }
```

````python
#Practical 2: List Comprehensions
#Q.1: https://www.hackerrank.com/challenges/list-comprehensions/problem

x = int(input())
y = int(input())
z = int(input())
n = int(input())
    
result=[[i,j,k] for i in range(x+1)
                    for j in range(y+1)
                    for k in range(z+1)
                    if (i+j+k)!=n]
print(result)
````

## Question 2
**Problem Statement:** Perform tuple operations to store product information and demonstrate packing, unpacking, concatenation, slicing, searching, and immutability.

````python
# Practical 2: 
# Q.2: Tuple operations for product information
# to use tuples to store immutable product info 
# create tuples containing product id,name,price

# tuple packing
product = (101, "Laptop", 50000)
print("Packed tuple:", product)

# tuple indexing
print("Product ID:", product[0])
print("Product Name:", product[1])
print("Product Price:", product[2])

# tuple unpacking
pid, pname, pprice = product
print("Unpacked tuple:", pid, pname, pprice)

#tuple concatenation
product2 = (102, "Smartphone", 30000)
product3 = (103, "Laptop", 45000)
combined_products = product + product2 + product3
print("Combined Products:", combined_products)

# count occurence of a value
print("Count of laptop:",combined_products.count('Laptop'))

# slicing
print("Sliced Tuple:", combined_products[0:3])

# indexing
print("Index of pid 103:",combined_products.index(103))


# immutability demo
try:
    product[2] = 55000
except TypeError as e:
    print("\nTuple is immutable:", e)

````

````python
# Practical 2: 
# Q.3: Set operations on customer ids
# objective: perform operations to identify common and unique customers
# question: given 2 sets of customer ids from 2 stores, perform: 
# union, intersection, difference, symmetric difference, add and remove an element

s1 = {101, 102, 103, 104, 105}
s2 = {104, 105, 106, 107, 108}

print("Union:", s1.union(s2))

print("Intersection:", s1.intersection(s2))

print("Difference:", s1.difference(s2))

print("Symmetric Difference:", s1.symmetric_difference(s2))

s1.add(109)
s1.remove(101)
print("Updated Set 1:", s1)
````

````python
# practical 2
#Q.6: Exception handling - Division calculator

try:
    num1 = int(input("Enter numerator: "))
    num2 = int(input("Enter denominator: "))
    result = num1 / num2
    print("Result:", result)
except ValueError:
    print("Please enter valid integers.")
except ZeroDivisionError:
    print("Denominator cannot be zero.")
````
