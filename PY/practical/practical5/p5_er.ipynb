# ER Diagram – Retail Orders Dataset

The ER diagram represents the structure of the Retail Orders dataset using four entities: Customer, Orders, Order Line and Category.

A customer can place multiple orders, and each order can contain multiple order lines. Each order line belongs to a particular product category.

The Order Line entity uses a composite primary key consisting of order_id and line_no to uniquely identify individual order lines. Foreign keys establish relationships between the entities.

The ER diagram helps illustrate the database structure, primary keys, foreign keys and one-to-many relationships, supporting organized data storage and reducing redundancy.

```mermaid
erDiagram
    CUSTOMER ||--o{ ORDERS : places
    ORDERS ||--|{ ORDER_LINE : contains
    CATEGORY ||--o{ ORDER_LINE : classifies

    CUSTOMER {
        int customer_id PK
    }

    ORDERS {
        int order_id PK
        date ordered_at
        int customer_id FK
        string channel
        string country
    }

    ORDER_LINE {
        int order_id PK, FK
        int line_no PK
        string category FK
        int units
        float unit_price_usd
        float discount_pct
        float revenue_usd
        boolean returned
    }

    CATEGORY {
        string category PK
    }
```
