```python
# p3q1 basic oop student class
# obj: create a class and objects
# ques: create student class containing rollno, name, multiple subjects and marks. Create methods to accept data, display data, and calculate grades
class Student:
    def __init__(self,rollno,name):
        self.rollno = rollno
        self.name = name
        self.subjects = []
        self.marks = []
    
    def accept(self):
        self.rollno=input("Enter rollno:")
        self.name=input("Enter name:")
        no_of_subjects = int(input("Enter number of subjects:"))

        for sub in range(no_of_subjects):
            subject = input(f"Enter subject {sub + 1} name:")
            mark = float(input(f"Enter marks for {subject}:"))
            self.subjects.append(subject)
            self.marks.append(mark)

    def calculate_grade(self, mark):
        if mark >= 90:
            return 'A'
        elif mark >= 80:
            return 'B'
        elif mark >= 70:
            return 'C'
        elif mark >= 60:
            return 'D'
        else:
            return 'F'

    def display(self):
        print("Roll No:",self.rollno)
        print("Name:",self.name)
        for subject, mark in zip(self.subjects, self.marks):
            print("\nSubject:",subject)
            print("Marks:",mark)
            print("\nGrade:",self.calculate_grade(mark))

student1 = Student(0,"")
student1.accept()
student1.display()
```

```python
# p3q2 student record management system using classes
# Store student details, add students, search by roll number,
# display all students, and calculate average marks.

class Student:
    def __init__(self, rollno, name):
        self.rollno = rollno
        self.name = name
        self.subjects = []
        self.marks = []

    def add_subject_marks(self, subject, mark):
        self.subjects.append(subject)
        self.marks.append(mark)

    def average_marks(self):
        if not self.marks:
            return 0
        return sum(self.marks) / len(self.marks)

    def display(self):
        print("Roll No:", self.rollno)
        print("Name:", self.name)
        for subject, mark in zip(self.subjects, self.marks):
            print(subject, ":", mark)
        print("Average marks:", self.average_marks())


class StudentRecordSystem:
    def __init__(self):
        self.students = []

    def add_student(self):
        rollno = input("Enter roll number: ")

        if self.find_student(rollno) is not None:
            print("Student with this roll number already exists.")
            return

        name = input("Enter student name: ")
        student = Student(rollno, name)
        number_of_subjects = int(input("Enter number of subjects: "))

        for subject_number in range(number_of_subjects):
            subject = input(f"Enter subject {subject_number + 1} name: ")
            mark = float(input(f"Enter marks for {subject}: "))
            student.add_subject_marks(subject, mark)

        self.students.append(student)
        print("Student added successfully.")

    def find_student(self, rollno):
        for student in self.students:
            if student.rollno == rollno:
                return student
        return None

    def search_student(self):
        rollno = input("Enter roll number to search: ")
        student = self.find_student(rollno)

        if student is None:
            print("Student not found.")
        else:
            student.display()

    def display_all_students(self):
        if not self.students:
            print("No student records available.")
            return

        for student in self.students:
            student.display()
            print()

    def calculate_average_marks(self):
        all_marks = []
        for student in self.students:
            all_marks.extend(student.marks)

        if not all_marks:
            print("No marks available.")
        else:
            print("Average marks:", sum(all_marks) / len(all_marks))


record_system = StudentRecordSystem()

while True:
    print("\n1. Add a new student")
    print("2. Search a student")
    print("3. Display all students")
    print("4. Calculate average marks")
    print("5. Exit")

    choice = input("Enter your choice: ")

    if choice == "1":
        record_system.add_student()
    elif choice == "2":
        record_system.search_student()
    elif choice == "3":
        record_system.display_all_students()
    elif choice == "4":
        record_system.calculate_average_marks()
    elif choice == "5":
        print("Program ended.")
        break
    else:
        print("Invalid choice.")
```
