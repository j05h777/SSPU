```python
# p3q1 basic oop student class
# obj: create a class and objects
# ques: create student class containing rollno, name, multiple subjects and marks. Create methods to accept data, display data, and calculate grades
class Student:
    def __init__(self,rollno,name):
        self.rollno = rollno
        self.name = name
        self.subjects = []
        self.marks = []
    
    def accept(self):
        self.rollno=input("Enter rollno:")
        self.name=input("Enter name:")
        no_of_subjects = int(input("Enter number of subjects:"))

        for sub in range(no_of_subjects):
            subject = input(f"Enter subject {sub + 1} name:")
            mark = float(input(f"Enter marks for {subject}:"))
            self.subjects.append(subject)
            self.marks.append(mark)

    def calculate_grade(self, mark):
        if mark >= 90:
            return 'A'
        elif mark >= 80:
            return 'B'
        elif mark >= 70:
            return 'C'
        elif mark >= 60:
            return 'D'
        else:
            return 'F'

    def display(self):
        print("Roll No:",self.rollno)
        print("Name:",self.name)
        for subject, mark in zip(self.subjects, self.marks):
            print("\nSubject:",subject)
            print("Marks:",mark)
            print("\nGrade:",self.calculate_grade(mark))

student1 = Student(0,"")
student1.accept()
student1.display()
```
