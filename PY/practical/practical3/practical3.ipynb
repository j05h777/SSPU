```python
# Q1) Objective: Create a class and objects.
# P3Q1a: Create a Student class containing: Roll Number, Name, Marks. Create methods to: Accept  data, Display data, Calculate grade

class Student:
    def __init__(self,rollno,name):
        self.rollno = rollno
        self.name = name
        self.subjects = []
        self.marks = []
    
    def accept(self):
        self.rollno=input("Enter rollno:")
        self.name=input("Enter name:")
        no_of_subjects = int(input("Enter number of subjects:"))

        for sub in range(no_of_subjects):
            subject = input(f"Enter subject {sub + 1} name:")
            mark = float(input(f"Enter marks for {subject}:"))
            self.subjects.append(subject)
            self.marks.append(mark)

    def calculate_grade(self, mark):
        if mark >= 90:
            return 'A'
        elif mark >= 80:
            return 'B'
        elif mark >= 70:
            return 'C'
        elif mark >= 60:
            return 'D'
        else:
            return 'F'

    def display(self):
        print("Roll No:",self.rollno)
        print("Name:",self.name)
        for subject, mark in zip(self.subjects, self.marks):
            print("\nSubject:",subject)
            print("Marks:",mark)
            print("\nGrade:",self.calculate_grade(mark))

student1 = Student(0,"")
student1.accept()
student1.display()
```

```python
# P3Q1b: Create a product class with: Product name, Price, Quantity, Method to calculate total value
class product():
    def __init__(self,name,price,quantity):
        self.name=name
        self.price=price
        self.quantity=quantity

    def total_value(self):
        print(f"Total value of {self.name}: {self.price*self.quantity}")

p=product("laptop",10000,3)
p.total_value()
```

```python
import pandas as pd

# S3Q3) Create a E-commerce Analytics Pipeline:
# S3Q3a) Construct a transformation pipeline for e-commerce transactions.

transactions = pd.DataFrame({
    "transaction_id": [1, 2, 3, 4, 5],
    "customer": ["A", "B", "A", "C", "B"],
    "product": ["Laptop", "Mouse", "Keyboard", "Laptop", "Mouse"],
    "category": ["Electronics", "Accessories", "Accessories", "Electronics", "Accessories"],
    "quantity": [1, 2, 1, 1, 3],
    "unit_price": [60000, 800, 1500, 60000, 800],
    "discount": [0.10, 0.05, 0, 0.15, 0.10]
})

# Transformation pipeline
ecommerce = (
    transactions
    .drop_duplicates()
    .assign(
        gross_value=lambda df: df["quantity"] * df["unit_price"],
        discount_value=lambda df: df["quantity"] * df["unit_price"] * df["discount"],
        net_value=lambda df: (
            df["quantity"] * df["unit_price"] * (1 - df["discount"])
        )
    )
)

print("Processed Transactions:")
display(ecommerce)

sales_summary = (
    ecommerce.groupby("category", as_index=False)
    .agg(
        total_orders=("transaction_id", "count"),
        total_quantity=("quantity", "sum"),
        total_sales=("net_value", "sum")
    )
    .sort_values("total_sales", ascending=False)
)

print("Sales Summary:")
display(sales_summary)
```

```python
# S3Q3b) Design a data-wrangling workflow for university survey data.
survey = pd.DataFrame({
    "student_id": [101, 102, 103, 104, 105],
    "department": ["CS", "IT", "CS", "Mechanical", "IT"],
    "year": [1, 2, 3, 2, 1],
    "satisfaction": [4, 5, None, 3, 4],
    "study_hours": [5, 7, 6, None, 8],
    "recommend": ["Yes", "Yes", "No", "Yes", "Yes"]
})

# Clean, transform, and summarize survey data
survey_clean = survey.copy()

survey_clean["satisfaction"] = survey_clean["satisfaction"].fillna(survey_clean["satisfaction"].median())
survey_clean["study_hours"] = survey_clean["study_hours"].fillna(survey_clean["study_hours"].median())
survey_clean["satisfaction_level"] = survey_clean["satisfaction"].map(lambda score: "Satisfied" if score >= 4 else "Needs Improvement")

department_summary = (
    survey_clean.groupby("department", as_index=False)
    .agg(
        students=("student_id", "count"),
        average_satisfaction=("satisfaction", "mean"),
        average_study_hours=("study_hours", "mean"),
        recommendation_rate=("recommend", lambda x: (x == "Yes").mean() * 100)
    )
)

print("Cleaned Survey Data:")
display(survey_clean)

print("Department-wise Survey Summary:")
display(department_summary)

```
