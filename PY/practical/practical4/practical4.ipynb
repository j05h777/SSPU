## Q4.1: Inventory Stock Management

Question: Maintain a list of product stock quantities. 
- Increase the second item's stock by 10, 
- decrease the third item's stock by 4,
- identify items with fewer than 6 units, 
- and calculate total inventory.

```python
stock = [5, 12, 8, 4, 15]
print("Stock:", stock)

# Increase second item's stock by 10
stock[1] += 10

# Decrease third item's stock by 4
stock[2] -= 4

# Identify items with fewer than 6 units
low_stock = [qty for qty in stock if qty < 6]

# Calculate total inventory
total = sum(stock)

print("Updated Stock:", stock)
print("Items with fewer than 6 units:", low_stock)
print("Total Inventory:", total)
```


## Q4.2: Contact Deduplication

Question: A contact management system receives email addresses with inconsistent capitalization and extra spaces. 

Normalize them and identify unique addresses.

```python
# Email addresses with inconsistent formatting
emails = [
    " John@gmail.com ",
    "alice@yahoo.com",
    "JOHN@gmail.com",
    " Bob@gmail.com ",
    "alice@yahoo.com ",
    "bob@gmail.com"
]

# Normalize email addresses
normalized = [email.strip().lower() for email in emails]

# Identify unique email addresses
unique_emails = sorted(list(set(normalized)))

# Display results
print("Normalized Emails:", normalized)
print("Unique Emails:", unique_emails)
print("Total Unique Contacts:", len(unique_emails))
```
