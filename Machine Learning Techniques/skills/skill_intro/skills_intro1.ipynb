 Suppose an online shopping company has collected customer purchase data from different stores before using the data for customer analytics and ml, the data scientist noticed several problems:

- duplicate customer records
- missing product prices
- incorrect data entries

need to:
- calculate business insight
- identify laptop customers

```python
import numpy as np, pandas as pd
df=pd.read_csv("skills1a.csv")
df
```

```python
df.drop_duplicates(subset='customer',inplace=True)
df
```

```python
df["price"]=df["price"].fillna(df.groupby("product")["price"].transform("mean"))
df
```

```python
df=df[df["price"].isnull()==False]
df
```

```python
#Business Insights
print("Average Price:", round(df["price"].mean(),2))
print("Total Revenue:", round(df["price"].sum(),2))
print("Products Sold:\n",df["product"].value_counts())
```

```python
laptop_customers=df[df["product"]=="laptop"]
laptop_customers
```
