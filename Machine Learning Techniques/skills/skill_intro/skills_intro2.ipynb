```python
import pandas as pd
```

```python
df=pd.read_csv("skills1b.csv")
df
```

```python
df.drop_duplicates()
df
```

```python
df['price']=df['price'].fillna(df.groupby('product')['price'].transform('mean'))
df
```

```python
df=df[df["price"].isnull()==False]
df
```

```python
#q1. Which product is purchased most frequently?
df.groupby('product')['quantity'].sum().idxmax()
```

```python
#q2. How many laptops were sold?
int(df[df['product']=='laptop']['quantity'].sum())
```

```python
#q3.Which category has the highest number of products?
col=df.groupby('category')['product'].count().idxmax()
val=df.groupby('category')['product'].count().max()

print(col,val)
```

```python
#4. Find number of male and female customers?
df['gender'].value_counts()
```

```python
#5. Find average customer age? 
float(df['age'].mean())
```

```python
#6. Find youngest customer?
float(df['age'].min())
```

```python
#7. Calculate Total sale, average sales, highest sale
total=df['price'].sum()
avg=df['price'].mean()
highest=df['price'].max()
print("total sales:",total,"\naverage sales:",round(avg,2),"\nhighest sale:",highest)
```

```python
#8. Which city has highest number of customers?
df['city'].value_counts().idxmax()
```

```python
#9. Which city generated the highest sale?
col=df.groupby('city')['price'].max().idxmax()
val=df['price'].max()

print(col,val)
```
