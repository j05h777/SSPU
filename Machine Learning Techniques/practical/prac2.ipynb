```python
import pandas as pd
df=pd.DataFrame({
    "Quantity":[2,3,5],
    "Price":[100,200,150],
})

df['Total Price']=df['Quantity']*df['Price']
df
```

```python
df1=pd.DataFrame({"Salary":[25000,30000,35000,40000]})
df1['Normalized']=(df1.Salary-df1.Salary.min())/(df1.Salary.max()-df1.Salary.min())
df1
```

```python
df=pd.DataFrame({"Date":['2023-01-01','2023-01-02','2023-01-03']})
df['Date']=pd.to_datetime(df['Date'])
df['Day']=df['Date'].dt.day
df['Month']=df['Date'].dt.month
df['Year']=df['Date'].dt.year

df
```
