Suppose a company has employees with salaries 25000, 30000, 35000, NaN, 45000

do data preprocessing:
- handle missing values
- normalize values
- calculate average value
- filter criteria for above 30000

```python
import numpy as np
data=np.array([25000,30000,35500,np.nan,45000])
print("Original salary:",data)

mean=np.nanmean(data)
print("\nAverage salary:",mean)

#data1=np.nan_to_num(data,nan=mean)
data[np.isnan(data)]=mean
print("Salary after filling NaN values:",data)

#minmax scaling
normalised_data=(data-np.min(data))/(np.max(data)-np.min(data))
print("\nNormalised data:",normalised_data)

filtered_data=data[data>30000].astype(int)
print("\nFiltered data for salaries above 30k:",filtered_data)
```

```python
#z-score standardization
standardized_data=(data-mean)/np.std(data)
print(standardized_data)
```
