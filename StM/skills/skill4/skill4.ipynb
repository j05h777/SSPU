Q1. Website Visits

```python
from scipy.stats import poisson

# Given
lamb = 5

# Calculations
p_zero = poisson.pmf(0, lamb)
at_least = 1 - p_zero

# Results
print("Probability of at least one visit:")
print("P(X = 0) =", round(p_zero, 4))
print("P(X >= 1) =", round(at_least, 4))
```

Q2. Machine Breakdowns:

```python
from scipy.stats import poisson

lamb = 1
x = 2

exact = poisson.pmf(x, lamb)

print("Probability of exactly 2 breakdowns:")
print("P(X =", x, ") =", round(exact, 4))
```
