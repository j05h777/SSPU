```python
from scipy.stats import binom

# Example 1: Manufacturing defect rate
n = 8
p = 0.05

# At least one product is defective
at_least_1 = 1 - binom.cdf(0, n, p)

# Exactly 5 products are defective
exactly_5 = binom.pmf(5, n, p)

# At most 3 products are defective
at_most_3 = binom.cdf(3, n, p)

print("Example 1")
print("Probability of at least 1 defective product:",
      round(at_least_1, 4))
print("Probability of exactly 5 defective products:",
      round(exactly_5, 4))
print("Probability of at most 3 defective products:",
      round(at_most_3, 4))


# Example 2: Customer purchasing probability
n = 7
p = 0.4

# At most 2 customers purchase
at_most_2 = binom.cdf(2, n, p)

# Exactly 3 customers purchase
exactly_3 = binom.pmf(3, n, p)

# At least 3 customers purchase
at_least_3 = 1 - binom.cdf(2, n, p)

print("\nExample 2")
print("Probability of at most 2 customers purchasing:",
      round(at_most_2, 4))
print("Probability of exactly 3 customers purchasing:",
      round(exactly_3, 4))
print("Probability of at least 3 customers purchasing:",
      round(at_least_3, 4))


# Example 3: Customer satisfaction
n = 15
p = 0.8

# Exactly 12 customers are satisfied
exactly_12 = binom.pmf(12, n, p)

# At least 13 customers are satisfied
at_least_13 = 1 - binom.cdf(12, n, p)

# Maximum 11 customers are satisfied
at_most_11 = binom.cdf(11, n, p)

print("\nExample 3")
print("Probability of exactly 12 satisfied customers:",
      round(exactly_12, 4))
print("Probability of at least 13 satisfied customers:",
      round(at_least_13, 4))
print("Probability of at most 11 satisfied customers:",
      round(at_most_11, 4))
```
