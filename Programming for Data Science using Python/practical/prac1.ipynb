```python
#Eligibility Checker for Bsc and Msc
print("\t\tBSc and MSc Eligibility Checker")

bsc_required_cutoff=60
msc_required_cutoff=6
print("1. Bsc\n2. Msc\n3. exit") 
while True:
    course=int(input("Enter option: "))
    match course:
        case 1:
            percentage=float(input("Input 12th percentage: "))
            if percentage>=bsc_required_cutoff:
                print("Eligible for Bsc\n")
            else:
                print("Not eligible for Bsc\n")
        case 2:
            cgpa=float(input("Input graduation cgpa: "))
            if cgpa>=msc_required_cutoff:
                print("Eligible for Msc\n")
            else:
                print("Not eligible for Msc\n")
        case 3:
            print("exiting...")
            break
        case _:
            print("Invalid choice")
```

```python

#Password Validator
import re
print("\t\tPassword Validator")

password=input("Enter password: ")
pattern=r"^(?=.*[A-Z])(?=.*[a-z])(?=.*\d)(?=.*[@$!%*?&])[A-Za-z\d@$!%*?&]{8,}$"

if re.match(pattern,password):
    print("Strong password")
else:
    print("Weak password")
```

```python
#Simple Calculator
def add(a, b):
    return a + b

def subtract(a,b):
    return a-b

def multiply(a,b):
    return a*b

def divide(a,b):
    return a/b

print("\t\tSimple Calculator")

a=float(input("\nEnter First Number: "))
b=float(input("Enter Second Number: "))

print("\n1. Addition")
print("2. Subtraction")
print("3. Multiplication")
print("4. Division")
print("5. Exit")

while True:
    choice=int(input("\nEnter choice:"))
    match choice:
        case 1:
            result=add(a,b)
            print("result:",result)
        case 2:
            result=subtract(a,b)
            print("result:",result)
        case 3:
            result=multiply(a,b)
            print("result:",result)
        case 4:
            if b==0:
                print("Cannot divide by zero")
            else:
                result=divide(a,b)
                print("result:",result)     
        case 5:
            print("exiting...")
            break
        case _:
            print("Invalid choice")
```
