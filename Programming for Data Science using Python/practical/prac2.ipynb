```python
#wap to input name and dob, from dob calculate the age. if age>=18 then print eligibity for voting, driving
from datetime import date
name=input("enter name")
dob=input("enter dob (yyyy-mm-dd)")

year,month,day=map(int,dob.split("-"))
birth_date=date(year, month, day)
today=date.today()
age=today.year-birth_date.year
if(today.month,today.day)<(birth_date.month, birth_date.day):
    age-=1

print("Age:",age)

if age>=18:
    f1=input("are you an indian citizen?(y/n)")
    f2=input("are you a female?(y/n)")
    if f1 == "y":
        print("eligible for voting and drivers license")
        if f2 == "y":
            print("eligible for sukanya yojana")
    else:
        print("not eligible for voting and drivers license")
else:
    print("not eligible as not indian citizen")
```

```python
#wap a py program for a foodcourt where first it displays all food items and their rates available. 
#the user has to pass the item code and its quantity 
#it will execute and ask the user the requirement until the user presses x in the item code and quantity 0
print("\t\tMENU")
print("Item Code|Item\t\t|Price\n" \
      "101\t |Samosa\t|rs.15\n" \
      "102\t |Vada\t\t|rs.20\n" \
      "103\t |Vadapav\t|rs.25\n"\
      "104\t |Bhel Puri\t|rs.30\n")

while(True):
    itemcode=input("Enter itemcode (x to exit):")
    quantity=int(input("Enter quantity (0 to exit):"))
    if itemcode == "x":
        print("Exiting")
        break
    else:
        match itemcode:
            case "101": print("Quantity:",quantity,"\nTotal Price: Rs.",15*quantity)
            case "102": print("Quantity:",quantity,"\nTotal Price: Rs.",20*quantity)
            case "103": print("Quantity:",quantity,"\nTotal Price: Rs.",25*quantity)
            case "104": print("Quantity:",quantity,"\nTotal Price: Rs.",30*quantity)
            case _ : print("Invalid Entry")
```
