```python
print("\t\tBMI Calculator")

weight=float(input("Enter your weight in kg: "))
height=float(input("Enter your height in meters: "))

bmi=(weight/(height)**2)

if bmi<18.5:
    print("\nYour BMI is: ",round(bmi,2),"\nYou are underweight")
elif bmi>=18.5 and bmi<24.9:
    print("\nYour BMI is: ",round(bmi,2),"\nYou are normal weight")
elif bmi>=25 and bmi<29.9:
    print("\nYour BMI is: ",round(bmi,2),"\nYou are overweight")
else:
    print("\nYour BMI is: ",round(bmi,2),"\nYou are obese")
```

```python
print("\t\tElectricity Bill Calculator")

units_consumed=float(input("\nEnter electricity units (kWh) consumed in a month: "))

#using slab logic to calculate electricity bill
#rates per kWh between Rs.5 to Rs.15
'''
First 100 units      → Rs.5/kWh
Next 100 units       → Rs.7.50/kWh
Next 100 units       → Rs.12/kWh
Above 300 units      → Rs.15/kWh
'''

if units_consumed<=100:
    bill_amount=units_consumed*5 #first slab
elif units_consumed<=200:
    bill_amount=(100*5)+((units_consumed-100)*7.5) #second slab
elif units_consumed<=300:
    bill_amount=(100*5)+(100*7.5)+((units_consumed-200)*12) #third slab
else:
    bill_amount=(100*5)+(100*7.5)+(100*12)+((units_consumed-300)*15) #fourth slab

print("\nTotal electricity units consumed:", units_consumed, "kWh")
print("Total bill amount: Rs.", round(bill_amount, 2))
```

```python
print("\tStudent Grade Calculator")

print("\nenter marks obtained in 5 subjects(out of 100):")
subject1=float(input("enter marks for subject 1: "))
subject2=float(input("enter marks for subject 2: "))
subject3=float(input("enter marks for subject 3: "))
subject4=float(input("enter marks for subject 4: "))
subject5=float(input("enter marks for subject 5: "))

total_marks_obtained=subject1+subject2+subject3+subject4+subject5
percentage=(total_marks_obtained/500)*100

if percentage>=90:
    grade="A+"
elif percentage>=80:
    grade="A"
elif percentage>=70:
    grade="B"
elif percentage>=60:
    grade="C"
elif percentage>=50:
    grade="D"
else:
    grade="F"

print("\nTotal marks obtained:",total_marks_obtained,"/ 500")
print("Percentage:", round(percentage, 2), "%")
print("Grade:", grade)
```
